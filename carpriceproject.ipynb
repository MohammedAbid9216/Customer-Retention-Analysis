```python
# Task 9: Outlier Detection
```

```python
# 1. Detect outliers using Boxplots.
sns.boxplot(df)
plt.xticks(rotation=45)
```

```python
# 2. Detect outliers using the IQR method.
for i in df[num_col]:
    q1,q3=df[i].quantile([0.25,0.75])
    iqr=q3-q1
    lower_bound=q1-1.5*iqr
    upper_bound=q3+1.5*iqr
    df_outlier=df[(df[i]<lower_bound) | (df[i]>upper_bound)]
    print(i,df_outlier)
```

```python

for i in df.columns:
  if df[i].dtype=='int64' or df[i].dtype=='float64':
    q1,q3=df[i].quantile([0.25,0.75])
    iqr=q3-q1
    lower_bound=q1-1.5*iqr
    upper_bound=q3+1.5*iqr
    df_outlier=df[(df[i]<lower_bound) | (df[i]>upper_bound)]
    print(i,df_outlier.shape)
```
